```python
def MergeHelper(L,R,i,j):
    if L[i] < R[j]:
        v = L[i]
        i = i +1
    else:
        v = R[j]
        j = j+1
    return i,j,v

def Merge(L,R,l,r):
    C = []
    i = 0
    j = 0
    l = len(L)
    r = len(R)
    while i < l and j < r:
        i,j,val = MergeHelper(L,R,i,j)
        C.append(val)

    while i < l:
        C.append(L[i])
        i = i+1

    while j < r:
        C.append(R[j])
        j = j+1
    return C

def MergeSort(A):
    n = len(A)
    if n <= 1:
        return A
    # low = 0 
    # high = n
    mid = (n)//2
    L = MergeSort(A[:mid])
    R = MergeSort(A[mid:])
    print(L)
    print(R)
    C = Merge(L,R,len(L),len(R))
    return C
```
